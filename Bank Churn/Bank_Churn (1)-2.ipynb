```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import tensorflow as tf
from sklearn import preprocessing
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score, f1_score, precision_recall_curve, auc
import matplotlib.pyplot as plt
from tensorflow.keras import optimizers

```

```python
project_path = '/content/drive/My Drive/Colab Notebooks/'
```

```python
dataset_file = project_path + 'bank.csv'
```

```python
data = pd.read_csv(dataset_file)
```

**DATA PREPROCESSING**

```python
data.head()
```

```python
data = data.drop(["RowNumber","CustomerId",'Surname'], axis = 1)
```

```python
X_data = data.iloc[:, :-1]
```

```python
print(X_data.shape)
print('')
X_data.head()
```

```python
y_data = data.iloc[:, -1]
```

```python
print(y_data.shape)
print('')
y_data.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_data, y_data, test_size = 0.3, random_state = 7)
X_train, x_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.25, random_state=1) # 0.25 x 0.8 = 0.2
```

```python
X_train.head()
```

```python
x_val.head()
```

```python
X_train = pd.get_dummies (X_train , columns = ['Geography','Gender'], dtype = 'float32', prefix = 'Country')
X_test = pd.get_dummies (X_test , columns = ['Geography','Gender'], dtype = 'float32', prefix = 'Country')
x_val = pd.get_dummies (x_val , columns = ['Geography','Gender'], dtype = 'float32', prefix = 'Country')
```

```python
X_train_norm = preprocessing.normalize(X_train)
x_val_norm = preprocessing.normalize(x_val)
x_val_norm = x_val_norm.astype('float32')
X_train_norm = X_train_norm.astype('float32')
```

```python
X_train_norm
```

```python
print(X_train_norm.shape)
print(x_val.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

**CREATING MODEL**

```python
model = Sequential()
```

Adding Layers

```python
model.add(Dense(13, input_shape = (13,) , activation = 'relu'))
model.add(Dense(100, activation = 'tanh'))
model.add(Dense(1, activation = 'sigmoid'))
```

```python
sgd = optimizers.Adam(lr = 0.001)
```

```python
model.compile(optimizer = sgd, loss = 'categorical_crossentropy', metrics=['accuracy'])
```

**Training the Model**

```python
model.fit(x_val_norm, y_val, batch_size = 200, epochs = 10, verbose = 1)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint as sp_randint
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
```

**Evaluation**

```python
X_test = preprocessing.normalize(X_test)
```

```python
results = model.evaluate(X_train, y_train.values)
```

With Binary crossentropy as the loss function I got a loss of .5, but with categorical the loss is significantly smaller at 2.4e-8. which is also smaller than the mse loss of .158. The number of epochs doesn't change the accuracy or loss nor does the bach size. I then changed the activation function on the hidden layer from tanh to relu and observed no changes to the results. The model was improved by one hot encoding the categorical values of geography and gender as well. Through all iterations the model accuracy was .8. I then trained the model on a validation set which had a higher validation accuracy than training accuracy which suggests overfitting or that there is a problem with the training data. After normalization the accuracy dropped further but the recall and precision values were more accurate with abetter F1 score of .31 instead of 0 suggesting a more generalizable model. Then I changed the activation functions of the input and the hidden layer to the default relu function which increased the accuracy from the new value of .6 to .8 again.
